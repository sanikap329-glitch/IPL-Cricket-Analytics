```python
import pandas as pd

column_names = [
    "record_type",
    "innings",
    "delivery_number",
    "batting_team",
    "batter",
    "non_striker",
    "bowler",
    "batter_runs",
    "total_extras",
    "wides",
    "noballs",
    "byes",
    "legbyes",
    "penalty_runs",
    "dismissal_method",
    "dismissed_players",
    "actual_delivery",
    "non_boundary",
    "fielder_1",
    "fielder_2",
    "fielder_3"
]

# Convert the ball records into a structured DataFrame
ball_df = pd.DataFrame(ball_records, columns=column_names)

# Replace empty strings with missing values
ball_df = ball_df.replace("", pd.NA)

# Convert numeric columns to appropriate types
numeric_columns = [
    "innings", "batter_runs", "total_extras",
    "wides", "noballs", "byes", "legbyes", "penalty_runs"
]

for column in numeric_columns:
    ball_df[column] = pd.to_numeric(
        ball_df[column], errors="coerce"
    ).fillna(0).astype(int)

# Calculate total runs scored on each delivery
ball_df["total_runs"] = (
    ball_df["batter_runs"] + ball_df["total_extras"]
)

# Count dismissed players on each delivery
ball_df["wickets"] = (
    ball_df["dismissed_players"]
    .fillna("")
    .apply(lambda x: len([p for p in x.split(",") if p.strip()]))
)

print("Dataset created successfully!")
print("Shape:", ball_df.shape)
display(ball_df.head(10))
```

```python
import os
import glob
import csv
import pandas as pd

# 1. Find all match CSV files
csv_files = glob.glob(
    os.path.join(data_folder, "**", "*.csv"),
    recursive=True
)

print("Total CSV files found:", len(csv_files))

# 2. Define the column names
column_names = [
    "record_type", "innings", "delivery_number",
    "batting_team", "batter", "non_striker", "bowler",
    "batter_runs", "total_extras", "wides", "noballs",
    "byes", "legbyes", "penalty_runs", "dismissal_method",
    "dismissed_players", "actual_delivery", "non_boundary",
    "fielder_1", "fielder_2", "fielder_3"
]

# 3. Read and combine ball records from every match
all_matches = []
failed_files = []

for file_path in csv_files:
    try:
        with open(file_path, "r", encoding="utf-8-sig", newline="") as file:
            records = list(csv.reader(file))

        match_balls = [
            row for row in records
            if row and row[0] == "ball"
        ]

        for row in match_balls:
            # Keep consistent field count
            row = row[:len(column_names)]
            row += [""] * (len(column_names) - len(row))

            all_matches.append({
                **dict(zip(column_names, row)),
                "source_file": os.path.basename(file_path)
            })

    except Exception as error:
        failed_files.append((file_path, str(error)))

# 4. Create the combined DataFrame
ipl_df = pd.DataFrame(all_matches)

print("\nCOMBINED DATASET")
print("-" * 40)
print("Total delivery records:", len(ipl_df))
print("Files processed:", len(csv_files) - len(failed_files))
print("Files with errors:", len(failed_files))
print("Dataset shape:", ipl_df.shape)

display(ipl_df.head())
```

```python
import os
import glob
import csv
import pandas as pd

# Match metadata fields stored in Cricsheet info records
metadata_keys = [
    "season",
    "date",
    "match_type",
    "venue",
    "city",
    "toss_winner",
    "toss_decision",
    "winner",
    "winner_runs",
    "winner_wickets",
    "player_of_match"
]

metadata_rows = []

for file_path in csv_files:
    try:
        with open(file_path, "r", encoding="utf-8-sig", newline="") as file:
            records = list(csv.reader(file))

        info = {}

        for row in records:
            if len(row) >= 3 and row[0] == "info":
                key = row[1]
                value = row[2]

                # Keep the first value for repeated metadata keys
                if key not in info:
                    info[key] = value

        metadata = {
            key: info.get(key, "")
            for key in metadata_keys
        }

        metadata["match_id"] = os.path.splitext(
            os.path.basename(file_path)
        )[0]

        metadata_rows.append(metadata)

    except Exception as error:
        print("Metadata error:", file_path, error)

match_metadata = pd.DataFrame(metadata_rows)

# Attach match information using the source filename
ipl_df["match_id"] = (
    ipl_df["source_file"].str.replace(".csv", "", regex=False)
)

ipl_df = ipl_df.merge(
    match_metadata,
    on="match_id",
    how="left",
    validate="many_to_one"
)

print("MATCH METADATA VALIDATION")
print("-" * 40)
print("Delivery records:", len(ipl_df))
print("Unique matches:", ipl_df["match_id"].nunique())
print("Seasons:", sorted(ipl_df["season"].dropna().unique()))
print("Unique batting teams:", ipl_df["batting_team"].nunique())

print("\nMissing metadata:")
print(
    ipl_df[metadata_keys]
    .replace("", pd.NA)
    .isna()
    .sum()
)

print("\nTeams in the dataset:")
print(sorted(ipl_df["batting_team"].dropna().unique()))

# Save the enriched dataset
enriched_path = os.path.join(
    output_folder,
    "ipl_all_matches_with_metadata.csv"
)

ipl_df.to_csv(enriched_path, index=False)

print("\nSaved enriched dataset to:")
print(enriched_path)
```

Create the team performance table


```python
import pandas as pd

# Create one row per match with its metadata
matches = (
    ipl_df.groupby("match_id", as_index=False)
    .agg(
        season=("season", "first"),
        team1=("batting_team", "first"),
        toss_winner=("toss_winner", "first"),
        toss_decision=("toss_decision", "first"),
        winner=("winner", "first")
    )
)

# Get both teams that participated in each match
teams_per_match = (
    ipl_df.groupby("match_id")["batting_team"]
    .agg(lambda x: list(pd.unique(x.dropna())))
)

matches["teams"] = matches["match_id"].map(teams_per_match)

# Count appearances and wins for each team
team_names = sorted({
    team
    for teams in matches["teams"]
    for team in teams
})

team_stats = []

for team in team_names:
    played = matches["teams"].apply(lambda teams: team in teams).sum()

    won = (
        (matches["winner"] == team)
        & matches["teams"].apply(lambda teams: team in teams)
    ).sum()

    team_stats.append({
        "Team": team,
        "Matches Played": int(played),
        "Matches Won": int(won),
        "Win Percentage": round(won / played * 100, 2) if played else 0
    })

team_performance = (
    pd.DataFrame(team_stats)
    .sort_values(
        by=["Win Percentage", "Matches Won"],
        ascending=False
    )
    .reset_index(drop=True)
)

display(team_performance)
```

```python
import matplotlib.pyplot as plt

wins_data = team_performance.sort_values("Matches Won")

plt.figure(figsize=(11, 7))
plt.barh(
    wins_data["Team"],
    wins_data["Matches Won"]
)

plt.title("IPL Team Performance: Total Matches Won")
plt.xlabel("Matches Won")
plt.ylabel("Team")
plt.tight_layout()
plt.show()
```

```python
import csv
import os

csv_path = r"C:\Users\sanik\OneDrive\Desktop\IPL_Cricket_Analytics\Data\outputs\ipl_all_matches_with_metadata.csv"

# Numeric columns that need conversion
integer_columns = {
    "innings", "batter_runs", "total_extras",
    "wides", "noballs", "byes", "legbyes",
    "penalty_runs", "winner_runs", "winner_wickets"
}

double_columns = {
    "delivery_number", "actual_delivery"
}

batch_size = 5000
batch = []
inserted_rows = 0

columns = [
    "record_type", "innings", "delivery_number",
    "batting_team", "batter", "non_striker", "bowler",
    "batter_runs", "total_extras", "wides", "noballs",
    "byes", "legbyes", "penalty_runs", "dismissal_method",
    "dismissed_players", "actual_delivery", "non_boundary",
    "fielder_1", "fielder_2", "fielder_3", "source_file",
    "match_id", "season", "date", "match_type", "venue",
    "city", "toss_winner", "toss_decision", "winner",
    "winner_runs", "winner_wickets", "player_of_match"
]

placeholders = ", ".join(["%s"] * len(columns))

insert_query = f"""
INSERT INTO ipl_deliveries_python
({", ".join(columns)})
VALUES ({placeholders})
"""

def convert_value(column, value):
    if value is None or value.strip() == "":
        return None

    if column in integer_columns:
        return int(float(value))

    if column in double_columns:
        return float(value)

    return value

with open(csv_path, "r", encoding="utf-8-sig", newline="") as file:

    reader = csv.DictReader(file)

    for row in reader:

        values = tuple(
            convert_value(col, row[col])
            for col in columns
        )

        batch.append(values)

        if len(batch) >= batch_size:
            cursor.executemany(insert_query, batch)
            connection.commit()

            inserted_rows += len(batch)
            print(f"Imported: {inserted_rows:,} rows")

            batch.clear()

    # Insert remaining records
    if batch:
        cursor.executemany(insert_query, batch)
        connection.commit()

        inserted_rows += len(batch)
        print(f"Imported: {inserted_rows:,} rows")

print("\nImport completed!")
print("Total inserted rows:", inserted_rows)
```
